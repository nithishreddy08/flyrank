```python
leakage_columns = [
    "trend_direction",
    "trend_pct",
    "is_declining_label"
]

found = [col for col in leakage_columns if col in df.columns]

print("Potential leakage columns:")
print(found)

print("\nAudit rule:")
print("trend_direction, trend_pct, and is_declining_label must NOT be features.")
```

I checked the final feature set for target leakage. The target-related variables trend_direction and trend_pct are excluded from the model features because the declining label is derived from trend information.

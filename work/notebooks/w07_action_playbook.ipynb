## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

1.Fix declining content first — Reason code: DECLINING. These pages are losing search trend and need attention first.

2.Improve low-performing content — Reason code: LOW_CTR. Pages getting impressions but few clicks may need better titles and descriptions.

3.Improve weak-position content — Reason code: LOW_RANK. Pages ranking poorly may need better content and search intent alignment.

4.Refresh outdated content — Reason code: REFRESH. Older content should be reviewed and updated when the data shows an opportunity.

The ranking prioritizes actions that can have the clearest impact while avoiding changes based only on missing or unreliable data.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

This playbook is for content and SEO teams to decide which pages should be reviewed first. It uses available search and content performance signals to create a practical action queue.

It is a decision-support tool, not an automatic publishing system. The recommendations should be checked by a human before making content changes.

The playbook is not valid when important data is missing, outdated, or unreliable. It should also not be used as the only source for major business decisions.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

A person must check the page content, search intent, business importance, data quality, and the reason for the recommendation before taking action.

The system should never automatically publish, delete, rewrite, or change important content. It should not make decisions when the required data is missing or unreliable. Final content and SEO decisions must remain with a human reviewer.

```python
review_checks = [
    "Check data quality and missing values",
    "Check search intent and page content",
    "Check business importance",
    "Check recommendation reason code",
    "Human approval required before action"
]

no_go_list = [
    "Do not auto-publish content",
    "Do not auto-delete content",
    "Do not make major SEO changes automatically",
    "Do not act when important data is missing or unreliable"
]

print("Human review checks:", len(review_checks))
print("No-go rules:", len(no_go_list))
print("Human approval required: YES")
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

The recommendations should be reviewed when search trends change, performance metrics shift significantly, or the data becomes stale.

A refresh or retraining should be considered when recommendation accuracy drops, the ranking of actions no longer matches human judgment, or new content types and search patterns appear. Missing or unusual data should also trigger a review.

```python
triggers = [
    "Search trends change significantly",
    "Performance metrics change significantly",
    "Recommendation accuracy drops",
    "Human reviewers disagree with recommendations",
    "Data becomes stale or unreliable",
    "New content types or search patterns appear"
]

print("Number of monitoring/retrain triggers:", len(triggers))
for i, trigger in enumerate(triggers, 1):
    print(f"{i}. {trigger}")

```
